```python
summary_rows = []

for sheet in xls.sheet_names:
    df = pd.read_excel(SOURCE_PATH, sheet_name=sheet, header=None)
    
    rows, cols = df.shape
    non_empty = df.notna().sum().sum()
    
    # 앞쪽 값 몇 개만 보기
    # 힌트: df.stack().dropna().head(10)
    preview_values = df.stack().dropna().head(10).tolist()
    
    summary_rows.append({
        "sheet": sheet,
        "rows": rows,
        "cols": cols,
        "non_empty_cells": non_empty,
        "preview_values": preview_values
    })

summary_df = pd.DataFrame(summary_rows)
summary_df
```
